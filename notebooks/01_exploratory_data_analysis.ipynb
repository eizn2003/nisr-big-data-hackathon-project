```python

# Cell 1: Import libraries for exploratory data analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display more columns when inspecting datasets
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

# Use a consistent style for our visualizations
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python

# Cell 2: Load the FinScope 2024 Rwanda dataset

from pathlib import Path

# Identify the project root
project_root = Path.cwd().parent

# Define the dataset path
data_path = project_root / "data" / "datasets" / "FinScope2024Rwanda.dta"

# Load the Stata dataset
df = pd.read_stata(data_path)

# Display basic information
print("Dataset loaded successfully!")
print(f"Number of observations: {df.shape[0]:,}")
print(f"Number of variables: {df.shape[1]:,}")

# Preview the first five observations
display(df.head())
```

```python

# Cell 3: Inspect the dataset structure

# Display the dataset's dimensions
print("Dataset shape:", df.shape)

# Display column names and data types
print("\nDataset information:")
df.info()

# Display the first 20 variable names
print("\nFirst 20 variable names:")
print(df.columns[:20].tolist())

# Count variables by data type
print("\nVariables by data type:")
print(df.dtypes.value_counts())
```

```python

# Cell 4: Check proposed variables against the dataset

# Variables identified in our initial data requirements
proposed_variables = [
    "a1", "a2", "a3", "a6",
    "pop_wt", "hh_wt", "hh_size", "hh_head",
    "b1", "b2", "c4a", "c4b", "c4d",
    "k1", "k2", "k3",
    "g3a", "g9", "g11a", "g11b",
    "h2a", "h3", "h7", "h8a", "h8b",
    "f4a", "f5", "f6",
    "l1", "l2",
    "i5b", "i6", "i7",
    "j1a", "j1b", "j3a", "j3b",
    "c1c", "c1e", "c5", "c6", "n1b"
]

# Check exact matches
available_variables = [
    var for var in proposed_variables if var in df.columns
]

missing_variables = [
    var for var in proposed_variables if var not in df.columns
]

print(f"Proposed variables checked: {len(proposed_variables)}")
print(f"Exact matches found: {len(available_variables)}")
print(f"Exact matches not found: {len(missing_variables)}")

print("\nEXACT MATCHES FOUND:")
print(available_variables)

print("\nEXACT MATCHES NOT FOUND:")
print(missing_variables)

# Search for related column names for variables not found exactly
print("\nRELATED VARIABLE NAMES:")

for var in missing_variables:
    related = [
        col for col in df.columns
        if col.lower().startswith(var.lower())
    ]

    if related:
        print(f"{var}: {related[:30]}")
```

```python

# Cell 5: Discover related variable names

# Prefixes representing key areas of our analysis
prefixes = [
    "a", "b", "c", "d", "f", "g", "h",
    "i", "j", "k", "l", "n"
]

for prefix in prefixes:
    matches = [
        col for col in df.columns
        if col.lower().startswith(prefix)
    ]

    print(f"\n--- Variables beginning with '{prefix}' "
          f"({len(matches)} found) ---")

    print(matches[:60])
```

```python
# Cell 6: Inspect the original survey question labels

reader = pd.io.stata.StataReader(data_path)
variable_labels = reader.variable_labels()

variables_to_check = [
    "a1", "a2", "a6",
    "b1", "b2",
    "c4a", "c4b", "c4d",
    "k1", "k2", "k3",
    "g3a", "h2", "h3",
    "f4a", "f5", "f6",
    "l1", "l2",
    "i5a", "i5b_1",
    "j1a", "j1b", "j3a", "j3b",
    "n1b"
]

for var in variables_to_check:
    if var in variable_labels:
        print(f"{var}: {variable_labels[var]}")
    else:
        print(f"{var}: Label not found")
```

```python

# Cell 7: Find text columns that may contain numbers

text_columns = [
    col for col in df.columns
    if pd.api.types.is_object_dtype(df[col])
    or pd.api.types.is_string_dtype(df[col])
    or isinstance(df[col].dtype, pd.CategoricalDtype)
]

print("Columns stored as text or categories:", len(text_columns))

# Focus first on columns stored as actual text
string_columns = [
    col for col in df.columns
    if (
        pd.api.types.is_object_dtype(df[col])
        or pd.api.types.is_string_dtype(df[col])
    )
]

print("Columns stored as text:", len(string_columns))

print("\nExamples of text-column values:")
for col in string_columns[:20]:
    print(f"{col}: {df[col].dropna().astype(str).unique()[:5]}")
```

```python

# Cell 8: Check whether text columns contain numeric-looking values

numeric_text_candidates = []

for col in string_columns:
    values = df[col].dropna().astype(str).str.strip()

    # Ignore empty strings when checking
    values = values[values != ""]

    if len(values) > 0:
        numeric_values = pd.to_numeric(values, errors="coerce")

        # If every non-empty value can be read as a number,
        # flag this column for further inspection
        if numeric_values.notna().all():
            numeric_text_candidates.append(col)

print(
    "Text columns containing only numeric-looking values:",
    len(numeric_text_candidates)
)
print("\nColumns to inspect:")
print(numeric_text_candidates)
```

```python

# Cell 9: Inspect storage types for key variables

variables_to_inspect = [
    "a1", "a2", "a6",
    "b1", "b2",
    "k1", "k2", "k3",
    "f4a", "f5", "f6",
    "hh_wt", "pop_wt"
]

for col in variables_to_inspect:
    if col in df.columns:
        print(f"\n{col}")
        print("Data type:", df[col].dtype)
        print("Sample values:")
        print(df[col].dropna().unique()[:8])
```

```python

# Cell 10: Check missing values in key variables

key_variables = [
    "a1", "a2", "a6",
    "b1", "b2", "c4a", "c4b", "c4d",
    "k1", "k2", "k3",
    "g3a", "h2", "h3",
    "f4a", "f5", "f6",
    "l1", "l2",
    "hh_wt", "pop_wt"
]

missing_report = pd.DataFrame({
    "data_type": df[key_variables].dtypes.astype(str),
    "missing_count": df[key_variables].isna().sum(),
    "missing_percent": (
        df[key_variables].isna().mean() * 100
    ).round(2),
    "unique_values": df[key_variables].nunique(dropna=True)
})

display(missing_report)
```

```python

# Cell 11: Investigate variable contents and possible skip patterns

for col in ["g3a", "h2", "h3"]:
    print(f"\n--- {col} ---")
    print("Data type:", df[col].dtype)
    print("First 5 non-missing values:")
    print(df[col].dropna().head().tolist())
    print("Number of distinct non-missing values:",
          df[col].nunique(dropna=True))

print("\n--- k1 versus missing k2 and k3 ---")
print(pd.crosstab(
    df["k1"],
    df["k2"].isna(),
    margins=True
))
print("\nk3 missingness by k1:")
print(pd.crosstab(
    df["k1"],
    df["k3"].isna(),
    margins=True
))

print("\n--- f5 versus missing f6 ---")
print(pd.crosstab(
    df["f5"],
    df["f6"].isna(),
    margins=True
))

print("\n--- l1 versus missing l2 ---")
print(pd.crosstab(
    df["l1"],
    df["l2"].isna(),
    margins=True
))
```

```python

# Cell 12: Count empty strings in text columns

empty_string_report = []

for col in string_columns:
    empty_count = (
        df[col]
        .eq("")
        .sum()
    )

    if empty_count > 0:
        empty_string_report.append({
            "variable": col,
            "empty_string_count": int(empty_count),
            "empty_string_percent": round(
                empty_count / len(df) * 100, 2
            )
        })

empty_string_report = pd.DataFrame(empty_string_report)

if not empty_string_report.empty:
    empty_string_report = empty_string_report.sort_values(
        "empty_string_count",
        ascending=False
    )
    display(empty_string_report.head(20))
    print(
        "\nTotal variables containing empty strings:",
        len(empty_string_report)
    )
else:
    print("No empty strings found.")
```

```python
# Cell 13: Identify variables that are completely empty

completely_empty = []

for col in df.columns:
    # Count empty strings in text columns
    if (
        pd.api.types.is_object_dtype(df[col])
        or pd.api.types.is_string_dtype(df[col])
    ):
        is_empty = df[col].isna() | df[col].eq("")
    else:
        is_empty = df[col].isna()

    if is_empty.all():
        completely_empty.append(col)

print("Total variables:", df.shape[1])
print("Completely empty variables:", len(completely_empty))
print(
    "Variables with at least some data:",
    df.shape[1] - len(completely_empty)
)

print("\nFirst 30 completely empty variables:")
print(completely_empty[:30])
```

```python
# Cell 14: Investigate the meaning of completely empty variables

reader = pd.io.stata.StataReader(data_path)
variable_labels = reader.variable_labels()

empty_variable_details = pd.DataFrame({
    "variable": completely_empty,
    "label": [
        variable_labels.get(col, "No label available")
        for col in completely_empty
    ]
})

display(empty_variable_details)
```

```python
# Cell 15: Display every empty variable and its full label

pd.set_option("display.max_rows", 150)
pd.set_option("display.max_colwidth", None)

display(
    empty_variable_details.sort_values("variable").reset_index(drop=True)
)
```

```python
# Cell 16: Check related variables for usable data

questions_to_check = [
    "c7c", "c7e", "c8e",
    "f7", "f8", "f10a",
    "g9", "g10", "g15",
    "h2", "h8a", "h8b",
    "i3", "i4b", "i5a", "i6",
    "k4", "k8a", "k8b", "k14",
    "p3", "p5"
]

for question in questions_to_check:
    # Find the question and related variables, e.g. h2, h2_1, h2_2
    related = [
        col for col in df.columns
        if col == question or col.startswith(question + "_")
    ]

    print(f"\n{'=' * 60}")
    print(f"Question family: {question}")

    for col in related:
        series = df[col]

        if (
            pd.api.types.is_object_dtype(series)
            or pd.api.types.is_string_dtype(series)
        ):
            valid = series.notna() & series.ne("")
        else:
            valid = series.notna()

        label = variable_labels.get(col, "No label")
        print(
            f"{col} | usable values: {valid.sum():,}/{len(df):,}"
            f" | label: {label}"
        )
```

```python
# Cell 17: Inspect values in selected response columns

sample_columns = [
    "f7_1", "f7_2",
    "h2_1", "h2_2", "h2_3",
    "k8a_1", "k8a_2",
    "g10_1", "g10_2",
    "p5_1", "p5_2"
]

for col in sample_columns:
    if col in df.columns:
        print(f"\n{col}")
        print("Label:", variable_labels.get(col, "No label"))
        print("Data type:", df[col].dtype)
        print("Values, including missing values:")
        print(df[col].value_counts(dropna=False).head(10))
```

```python
# Cell 18: Check mobile-money question patterns

columns_to_check = [
    "f5",
    "f6",
    "f7_1",
    "f7_2",
    "f8_1"
]

for col in columns_to_check:
    if col in df.columns:
        print(f"\n--- {col} ---")
        print("Question:", variable_labels.get(col, "No label"))
        print(df[col].value_counts(dropna=False).head(10))
```

```python
# Cell 28: Inspect actual savings responses

sample_columns = [
    "g3a_01",   # Commercial bank
    "g3a_03",   # Mobile money
    "g3a_08",   # Umurenge SACCO
    "g3a_11",   # Informal savings group
    "g3a_15",   # Cash hidden at home
    "g3a_17",   # Do not save
]

for col in sample_columns:
    print(f"\n--- {col} ---")
    print("Label:", variable_labels.get(col, "No label available"))
    display(df[col].value_counts(dropna=False))
```

```python
# Cell 29: Check whether "Do not save" overlaps
# with other reported savings mechanisms

other_savings_cols = [
    "g3a_01", "g3a_03", "g3a_08",
    "g3a_11", "g3a_15"
]

do_not_save = df["g3a_17"].eq("Yes")

has_other_saving = df[other_savings_cols].eq("Yes").any(axis=1)

print("Respondents selecting 'Do not save':",
      int(do_not_save.sum()))

print("Of those, also reporting at least one of the")
print("selected other savings mechanisms:",
      int((do_not_save & has_other_saving).sum()))

print("Respondents selecting 'Do not save' only among")
print("these six checked variables:",
      int((do_not_save & ~has_other_saving).sum()))
```

```python
# Cell 30: Check "Do not save" against all other mechanisms

other_savings_cols = [
    f"g3a_{i:02d}" for i in range(1, 17)
]

do_not_save = df["g3a_17"].eq("Yes")

has_any_other_saving = (
    df[other_savings_cols].eq("Yes").any(axis=1)
)

print("Respondents selecting 'Do not save':",
      int(do_not_save.sum()))

print("Also reporting another savings mechanism:",
      int((do_not_save & has_any_other_saving).sum()))

print("Do not save, with no other mechanism selected:",
      int((do_not_save & ~has_any_other_saving).sum()))
```

```python
# Cell 31: Inspect reasons for not saving

sample_columns = [
    "g9_1",   # Never thought about it
    "g9_2",   # Prefer spending on other needs
    "g9_9",   # Don't know about savings/investment
    "g9_10",  # Don't have money to save
    "g9_11",  # Don't have a bank account
    "g9_12",  # Too expensive
    "g9_13",  # Don't have a job
]

for col in sample_columns:
    print(f"\n--- {col} ---")
    print("Label:", variable_labels.get(col, "No label available"))
    display(df[col].value_counts(dropna=False))
```

```python
# Cell 32: Compare saving status with g9 missingness

g9_missing = df["g9_10"].isna()
do_not_save = df["g3a_17"].eq("Yes")

print("Saving status vs g9_10 missingness:")

display(pd.crosstab(
    do_not_save,
    g9_missing,
    rownames=["Selected 'Do not save'"],
    colnames=["g9_10 is missing"],
    margins=True
))
```

```python
# Cell 33: Examine respondent age

print("Age variable label:")
print(variable_labels.get("b1", "No label available"))

print("\nAge summary:")
display(df["b1"].describe())

print("\nNumber of respondents by age:")
display(df["b1"].value_counts().sort_index())
```

```python
# Cell 34: Summarize age and create age groups

print("Age summary:")
display(df["b1"].describe())

# Create broad age groups
age_bins = [15, 24, 34, 44, 54, 64, 120]
age_labels = [
    "16–24",
    "25–34",
    "35–44",
    "45–54",
    "55–64",
    "65+"
]

df["age_group"] = pd.cut(
    df["b1"],
    bins=age_bins,
    labels=age_labels
)

print("\nRespondents by age group:")
display(df["age_group"].value_counts().sort_index())

print("\nAge group percentages:")
display(
    df["age_group"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
# Cell 35: Examine respondent sex

print("Sex distribution:")
display(df["b2"].value_counts())

print("\nSex distribution percentages:")
display(
    df["b2"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Cell 36: Examine education levels

print("Education levels:")
display(df["c4a"].value_counts(dropna=False))

print("\nEducation level percentages:")
display(
    df["c4a"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python
# Cell 37: Education level vs bank-account ownership

education_bank = pd.crosstab(
    df["c4a"],
    df["k1"],
    normalize="index"
).mul(100).round(2)

print("Percentage within each education level:")
display(education_bank)
```

```python
# Cell 38: Inspect mobile-money experience

print("Mobile-money experience categories:")
display(df["f5"].value_counts(dropna=False))

print("\nMobile-money experience percentages:")
display(
    df["f5"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python
# Cell 39: Education level vs mobile-money experience

mobile_money_by_education = pd.crosstab(
    df["c4a"],
    df["f5"],
    normalize="index"
).mul(100).round(2)

display(mobile_money_by_education)
```

```python
# Cell 40: Inspect reasons for limited transaction-account use

for i in range(1, 14):
    col = f"k8a_{i}"

    print(f"\n{col}: {variable_labels.get(col, 'No label available')}")
    display(df[col].value_counts(dropna=False))
```

```python
# Cell 41: Check who has valid k8a responses

has_k8a_response = df["k8a_7"].notna()

print("k1 responses among people with a valid k8a_7 answer:")
display(df.loc[has_k8a_response, "k1"].value_counts(dropna=False))

print("\nk1 responses among people with a missing k8a_7 answer:")
display(df.loc[~has_k8a_response, "k1"].value_counts(dropna=False))

print("\nCross-tabulation:")
display(pd.crosstab(
    df["k1"],
    has_k8a_response,
    rownames=["Own bank account"],
    colnames=["Has valid k8a_7 response"]
))
```

```python
# Cell 42: Examine borrowing and credit activity

credit_variables = {
    "h2_1": "Borrowed money in the past 12 months",
    "h2_2": "Repaid borrowed money",
    "h2_3": "Received goods/services and paid later"
}

for col, description in credit_variables.items():
    print(f"\n{description}")

    counts = df[col].value_counts(dropna=False)
    percentages = (
        df[col].value_counts(normalize=True, dropna=False) * 100
    ).round(2)

    display(pd.DataFrame({
        "Count": counts,
        "Percentage": percentages
    }))
```

```python
# Cell 43: Compare reported non-saving by urban/rural location

saving_by_location = pd.crosstab(
    df["a6"],
    df["g3a_17"],
    normalize="index"
).mul(100).round(2)

print("Savings status by urban/rural location (% within each group):")
display(saving_by_location)
```

```python
# Cell 44: Summarize key financial-inclusion indicators

key_indicators = {
    "Own bank account": (df["k1"] == "Yes"),
    "Registered mobile-money user": (
        df["f5"] == "You are a registered mobile money user"
    ),
    "Selected 'Do not save'": (df["g3a_17"] == "Yes"),
    "Borrowed in past 12 months": (df["h2_1"] == "Yes"),
}

summary_rows = []

for indicator, condition in key_indicators.items():
    summary_rows.append({
        "Indicator": indicator,
        "Respondents (Yes)": int(condition.sum()),
        "Sample percentage (%)": round(condition.mean() * 100, 2)
    })

key_indicator_summary = pd.DataFrame(summary_rows)
display(key_indicator_summary)
```

```python
# Cell 45: Plot key financial-inclusion indicators

import matplotlib.pyplot as plt

plot_data = key_indicator_summary.sort_values(
    "Sample percentage (%)",
    ascending=True
)

plt.figure(figsize=(9, 5))

plt.barh(
    plot_data["Indicator"],
    plot_data["Sample percentage (%)"]
)

plt.xlabel("Respondents (%)")
plt.title("Key Financial-Inclusion Indicators — FinScope 2024")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```
